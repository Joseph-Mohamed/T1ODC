## Normalization (Min-Max Scaling)
***Mathematical Formula:***

X' = (X - Xmin) / (Xmax - Xmin)

It rescales data into a fixed range, usually:

0 ≤ X' ≤ 1

***Use it when:***
- Data has a known minimum and maximum.
- Features need to be on the same scale.
- Using distance-based algorithms such as KNN or K-Means.
- No significant outliers are present.


# **Standardization (Z-Score Scaling)**
***Mathematical Formula:***

Z = (X - μ) / σ

***Where:***

X = Original value

μ = Mean

σ = Standard deviation

***It transforms the data so that:***
- Mean = 0
- Standard Deviation = 1
image.png
***Use it when:***
- Data is approximately normally distributed.
- Features have different units or scales.
- Using algorithms such as Logistic Regression.
- The dataset contains outliers, since it is generally less affected by them than Min-Max Scaling.
